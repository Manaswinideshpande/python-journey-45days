```python
nums= [3,2,4,5,6,73,3,3,33,6]
for i in nums:
    if i == 3:
        nums.remove(i)
print(nums)
```

```python
nums= [3,2,4,5,6,73,3,3,33,6]
emptylist=[]
for i in nums:
    if i != 3:
        emptylist.append(i)
print(emptylist)
```

```python
print([i for i in nums if i!=3])
```
