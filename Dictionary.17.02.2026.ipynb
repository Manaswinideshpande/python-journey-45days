```python
dictionary = {}

while True:
    user_name = input("Enter username: ")
    password = input("Enter password: ")

    dictionary[user_name] = password

    stop = input("Do you want to add more? (yes/no): ")

    if stop.lower() == "no":
        break

print("Final Dictionary:")
print(dictionary)
```
