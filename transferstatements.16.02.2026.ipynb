```python
### continue code is used to skip the rest of the code inside a loop
# if condition:
    # continue 

for i in range(13):
    if i ==3: # continue skips this particular element here it is 3
        continue
    print(i)
    
print('after continue values of i',i)
    
```

```python
### manufacturing units uses this continue
products = ['ok','defect','defect','ok','ok','ok','ok','ok','ok','ok','ok','ok','ok','defect','defect','defect']
for i in products:
    if i == 'defect':
        continue
    print(i)
```
